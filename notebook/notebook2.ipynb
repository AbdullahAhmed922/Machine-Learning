```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
# The dataset is located in the 'data' folder relative to the notebook
df = pd.read_csv("./data/salary_dataset.csv")
```

```python
print(df.head())    
print(df.columns)    
```

```python
X = df[["YearsExperience"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict(pd.DataFrame({"YearsExperience": [12]}))
print(f"Predicted Salary for 12 years of experience: {prediction[0]:.2f}")

m = model.coef_[0]
c = model.intercept_
print(f"\nThe linear regression equation is: Salary = {m:.2f} * YearsExperience + {c:.2f}")
```
